# 2. Create raw-data manifests

Create video and camera-name CSV files from the data staged by notebook 1.

```python
import csv
from pathlib import Path

PROJECT_ROOT = next(
    parent
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "notebooks").is_dir() and (parent / "annotation_pipeline").is_dir()
)
DATA_DIR = PROJECT_ROOT / "data" / "raw"
clip_list = DATA_DIR / "annotations" / "kitware" / "list-of-annotated-meva-clips.txt"
metadata_dir = DATA_DIR / "metadata"

clip_names = [
    line.split("|", maxsplit=1)[-1].strip()
    for line in clip_list.read_text(encoding="utf-8").splitlines()[2:]
    if line.strip()
]
with (DATA_DIR / "video_names.csv").open("w", newline="", encoding="utf-8") as output:
    writer = csv.DictWriter(output, fieldnames=["video_name"])
    writer.writeheader()
    writer.writerows({"video_name": clip_name} for clip_name in clip_names)

camera_rows = [
    {"metadata_file": file.name, "camera_name": file.stem.split(".")[-1]}
    for file in sorted(metadata_dir.glob("*.krtd"))
]
with (DATA_DIR / "camera_names.csv").open("w", newline="", encoding="utf-8") as output:
    writer = csv.DictWriter(output, fieldnames=["metadata_file", "camera_name"])
    writer.writeheader()
    writer.writerows(camera_rows)

print(f"Wrote {len(clip_names)} video names and {len(camera_rows)} camera rows.")
```
